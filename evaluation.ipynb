#coherence

````python
import pandas as pd
import ast
import numpy as np
from scipy.stats import kendalltau

# =========================
# 1. Read CSV
# =========================

df = pd.read_csv(importance_score_metrics)

# 10 columns containing the perturbation scores
perturbation_columns = df.columns.tolist()

print("Number of perturbation columns:", len(perturbation_columns))
print("Columns:", perturbation_columns)


# =========================
# 2. Parse arrays
# =========================


def parse_array(value):
    if isinstance(value, list):
        return value

    try:
        return ast.literal_eval(str(value))
    except:
        return None


# =========================
# 3. Calculate Kendall Tau
# =========================

all_results = []

for row_idx, row in df.iterrows():

    # Store the 3-feature importance scores
    scores = {}

    for col in perturbation_columns:

        values = parse_array(row[col])

        if values is not None and len(values) == 3:
            scores[col] = np.array(values, dtype=float)

    # Compare every pair of perturbations
    for i in range(len(perturbation_columns)):

        for j in range(i + 1, len(perturbation_columns)):

            col1 = perturbation_columns[i]
            col2 = perturbation_columns[j]

            if col1 not in scores or col2 not in scores:
                continue

            # Rank of the 3 features
            rank1 = pd.Series(scores[col1]).rank(method="average").values
            rank2 = pd.Series(scores[col2]).rank(method="average").values

            tau, p_value = kendalltau(rank1, rank2)

            all_results.append(
                {
                    "row": row_idx,
                    "perturbation_1": col1,
                    "perturbation_2": col2,
                    "kendall_tau": tau,
                    "p_value": p_value,
                }
            )


# Convert results to DataFrame
results_df = pd.DataFrame(all_results)

# Save pairwise results
results_df.to_csv("kendall_tau_pairwise_codebleu.csv", index=False)


# =========================
# 4. Overall statistics
# =========================

print("\n==============================")
print("Kendall Tau Results")
print("==============================")

print("Number of comparisons:", len(results_df))

print("\nMean Kendall Tau:")
print(results_df["kendall_tau"].mean())

print("\nMedian Kendall Tau:")
print(results_df["kendall_tau"].median())

print("\nStandard Deviation:")
print(results_df["kendall_tau"].std())


# =========================
# 5. Percentage of stable rankings
# =========================

positive = (results_df["kendall_tau"] > 0).mean() * 100
perfect = (results_df["kendall_tau"] == 1).mean() * 100

print("\nPositive Kendall Tau (%):", round(positive, 2))
print("Perfect agreement (%):", round(perfect, 2))
````

````python
import pandas as pd
import ast
import re
import time
from typing import Any, Dict

# ============================================================
# Configuration
# ============================================================

INPUT_FILE = senstivity_output
OUTPUT_FILE = "output_with_test_results.csv"

TEST_COLUMN = "state['test']"
CODE_COLUMN = "ablatedd_code_phi"
ERROR_COLUMN = 'state["error_message"]'

RESULT_COLUMN = "ablatedd_test_result"
COMPARISON_COLUMN = "ablatedd_error_comparison"


# ============================================================
# Read the CSV file
# ============================================================

df = pd.read_csv(INPUT_FILE, encoding="utf-8-sig")


# ============================================================
# Extract Python code from Markdown
# ============================================================


def extract_python_code(markdown_text: str):
    """Extract the code inside a ```python ... ``` block, if present.

    Return the original text if no code block is found."""

    if markdown_text is None:
        return ""

    markdown_text = str(markdown_text)

    pattern = r"```python\s*(.*?)```"

    match = re.search(pattern, markdown_text, re.DOTALL | re.IGNORECASE)

    if match:
        return match.group(1).strip()

    return markdown_text.strip()


# ============================================================
# Extract the main function name
# ============================================================


def extract_function_signature(source_code):
    """Extract the name of the first function defined in the candidate code."""

    try:

        source_code = extract_python_code(source_code)

        tree = ast.parse(source_code)

        for node in ast.walk(tree):

            if isinstance(node, ast.FunctionDef):

                args = node.args

                param_parts = []

                # Positional-only arguments
                for arg in args.posonlyargs:
                    param_parts.append(arg.arg)

                # Regular arguments
                for arg in args.args:
                    param_parts.append(arg.arg)

                # *args
                if args.vararg:
                    param_parts.append("*" + args.vararg.arg)

                # Keyword-only arguments
                for arg in args.kwonlyargs:
                    param_parts.append(arg.arg)

                # **kwargs
                if args.kwarg:
                    param_parts.append("**" + args.kwarg.arg)

                params_str = ", ".join(param_parts)

                signature = f"def {node.name}({params_str}):"

                return node.name, signature

        return None, None

    except SyntaxError:

        return None, None

    except Exception:

        return None, None


# ============================================================
# Convert a three-element array string to a list
# ============================================================


def parse_code_array(value):
    """Convert the ablatedd_code_phi column value to a list.

    Example: "['code1', 'code2', 'code3']" becomes
    ['code1', 'code2', 'code3']."""

    if pd.isna(value):
        return []

    try:

        parsed = ast.literal_eval(str(value))

        if isinstance(parsed, (list, tuple)):

            return list(parsed)

        return []

    except Exception as e:

        print("Error parsing code array:", e)

        return []


# ============================================================
# CodeTester
# ============================================================


class CodeTester:
    """Load the generated code and test it using the check function."""

    def __init__(self, generated_code: str, check_code: str, entry_point: str):

        self.generated_code = generated_code
        self.check_code = check_code
        self.entry_point = entry_point

    # --------------------------------------------------------
    # Load candidate function
    # --------------------------------------------------------

    def load_function_from_code(self) -> Any:

        generated_code = extract_python_code(self.generated_code)

        ns: Dict[str, Any] = {"__name__": "__generated__"}

        # Execute the candidate code
        exec(generated_code, ns, ns)

        func = ns.get(self.entry_point)

        if not callable(func):

            raise RuntimeError(
                f"Entry point '{self.entry_point}' " f"not found or not callable."
            )

        return func

    # --------------------------------------------------------
    # Load check function
    # --------------------------------------------------------

    def load_check_from_string(self) -> Any:

        check_code = extract_python_code(self.check_code)

        ns: Dict[str, Any] = {"__name__": "__check__"}

        exec(check_code, ns, ns)

        check_fn = ns.get("check")

        if not callable(check_fn):

            raise RuntimeError(
                "Function 'check' not found " "or not callable in check string."
            )

        return check_fn

    # --------------------------------------------------------
    # Run tests
    # --------------------------------------------------------

    def run_tests(self) -> bool:

        try:

            # Load candidate
            candidate = self.load_function_from_code()

            # Load check
            check_fn = self.load_check_from_string()

            # Execute all assertions
            check_fn(candidate)

            # Reaching this point means all tests passed
            return True

        except AssertionError:

            # At least one assertion failed
            return False

        except Exception as e:

            # Syntax, import, runtime, or other error
            return False

    # --------------------------------------------------------
    # Run and time
    # --------------------------------------------------------

    def run_and_time(self):

        start_time = time.time()

        result = self.run_tests()

        end_time = time.time()

        execution_time = end_time - start_time

        return result, execution_time


# ============================================================
# Convert error_message to True or False
# ============================================================


def get_expected_result(error_message):
    """Return True if error_message is empty.
    Return False if an error message is present."""

    # NaN
    if pd.isna(error_message):
        return True

    value = str(error_message).strip()

    # Empty
    if value == "":
        return True

    # Handle boolean values directly
    if value.lower() == "true":
        return True

    if value.lower() == "false":
        return False

    # Any nonempty error message means False
    return False


# ============================================================
# Process the dataset
# ============================================================

all_test_results = []

all_comparison_results = []


for index, row in df.iterrows():

    print(f"Processing row " f"{index + 1}/{len(df)}")

    # ========================================================
    # Get the test code
    # ========================================================

    test_code = row[TEST_COLUMN]

    if pd.isna(test_code):

        test_code = ""

    test_code = str(test_code)

    # ========================================================
    # Get the three candidate code snippets
    # ========================================================

    candidate_codes = parse_code_array(row[CODE_COLUMN])

    # ========================================================
    # Results for the three candidates
    # ========================================================

    test_results = []

    # ========================================================
    # Execute each candidate
    # ========================================================

    for candidate_index, candidate_code in enumerate(candidate_codes):

        print(f"    Candidate " f"{candidate_index + 1}/3")

        # ----------------------------------------------------
        # Extract the function name
        # ----------------------------------------------------

        entry_point, signature = extract_function_signature(candidate_code)

        # ----------------------------------------------------
        # Handle a missing function
        # ----------------------------------------------------

        if entry_point is None:

            print("    Could not find function.")

            test_results.append(False)

            continue

        # ----------------------------------------------------
        # Create the tester
        # ----------------------------------------------------

        tester = CodeTester(
            generated_code=candidate_code, check_code=test_code, entry_point=entry_point
        )

        # ----------------------------------------------------
        # Run the test
        # ----------------------------------------------------

        result = tester.run_tests()

        test_results.append(result)

        print(f"    Result: {result}")

    # ========================================================
    # Ensure there are exactly three results
    # ========================================================

    while len(test_results) < 3:

        test_results.append(False)

    if len(test_results) > 3:

        test_results = test_results[:3]

    # ========================================================
    # Store the actual test results
    #
    # Example:
    #
    # [True, False, False]
    # ========================================================

    all_test_results.append(test_results)

    # ========================================================
    # Expected result from error_message
    # ========================================================

    expected_result = get_expected_result(row[ERROR_COLUMN])

    # ========================================================
    # Compare the results
    #
    # Equal     -> 0
    # Different -> 1
    # ========================================================

    comparison_results = []

    for actual_result in test_results:

        if actual_result == expected_result:

            comparison_results.append(0)

        else:

            comparison_results.append(1)

    # ========================================================
    # Store the comparison
    #
    # Example:
    #
    # [1, 0, 1]
    # ========================================================

    all_comparison_results.append(comparison_results)


# ============================================================
# Add the columns
# ============================================================

df[RESULT_COLUMN] = all_test_results

df[COMPARISON_COLUMN] = all_comparison_results


# ============================================================
# Save the file
# ============================================================

df.to_csv(OUTPUT_FILE, index=False, encoding="utf-8-sig")


# ============================================================
# Summary
# ============================================================

print()
print("=" * 60)
print("Processing completed!")
print("=" * 60)

print(f"Input file : {INPUT_FILE}")

print(f"Output file: {OUTPUT_FILE}")

print()
print(df[[CODE_COLUMN, ERROR_COLUMN, RESULT_COLUMN, COMPARISON_COLUMN]].head())
````

````python
import pandas as pd
import ast

# ============================================================
# Configuration
# ============================================================

INPUT_FILE = senstivity_output
OUTPUT_FILE = "feature_discrimination_results.csv"

# METHODS = [
#     "sim_chrff_phi",
#     "embedd_lower_phi",
#     "codebleu_phi"
# ]

METHODS = ["  sim_chrff_phi", "  embedd_lower_phi", "  codebleu_phi"]

COMPARISON_COLUMN = "ablatedd_error_comparison"


# ============================================================
# Read the CSV file
# ============================================================

df = pd.read_csv(INPUT_FILE, encoding="utf-8-sig")


# ============================================================
# Convert an array string to a list
# ============================================================


def parse_array(value):
    """Convert an array string to a list.

    Example: "[0.12, 0.35, 0.21]" becomes [0.12, 0.35, 0.21]."""

    if pd.isna(value):
        return None

    try:
        result = ast.literal_eval(str(value))

        if isinstance(result, (list, tuple)):
            return list(result)

        return None

    except Exception as e:
        return None


# ============================================================
# Compute the result for each method
# ============================================================

for method in METHODS:

    results = []

    for index, row in df.iterrows():

        # ----------------------------------------------------
        # Importance scores for the three features
        # ----------------------------------------------------

        importance_scores = parse_array(row[method])

        # ----------------------------------------------------
        # Values from ablatedd_error_comparison
        # ----------------------------------------------------

        comparison = parse_array(row[COMPARISON_COLUMN])

        # ----------------------------------------------------
        # Validate the data
        # ----------------------------------------------------

        if (
            importance_scores is None
            or comparison is None
            or len(importance_scores) != 3
            or len(comparison) != 3
        ):

            results.append("Invalid")
            continue

        # ----------------------------------------------------
        # Find the index of the highest score
        # ----------------------------------------------------

        max_index = importance_scores.index(max(importance_scores))

        # ----------------------------------------------------
        # Comparison value corresponding to
        # the most important feature
        # ----------------------------------------------------

        corresponding_value = comparison[max_index]

        # ----------------------------------------------------
        # If the corresponding value is 1:
        # Correct
        #
        # Otherwise:
        # Incorrect
        # ----------------------------------------------------

        if corresponding_value == 1:

            results.append("Correct")

        else:

            results.append("Incorrect")

    # --------------------------------------------------------
    # Store the result for this method
    # --------------------------------------------------------

    df[f"{method}_discrimination"] = results


# ============================================================
# Calculate accuracy for each method
# ============================================================

summary = []


for method in METHODS:

    result_column = f"{method}_discrimination"

    valid_results = df[df[result_column].isin(["Correct", "Incorrect"])][result_column]

    correct = (valid_results == "Correct").sum()

    total = len(valid_results)

    if total > 0:

        accuracy = correct / total

    else:

        accuracy = 0

    summary.append(
        {"method": method, "correct": correct, "total": total, "accuracy": accuracy}
    )


# ============================================================
# Build the summary table
# ============================================================

summary_df = pd.DataFrame(summary)


# Accuracy as a percentage
summary_df["accuracy_percent"] = summary_df["accuracy"] * 100


# ============================================================
# Save the results file
# ============================================================

df.to_csv(OUTPUT_FILE, index=False, encoding="utf-8-sig")


# ============================================================
# Display the results
# ============================================================

print("\n==============================================")
print("Feature Discrimination Results")
print("==============================================\n")

print(summary_df.to_string(index=False))


print("\n==============================================")
print(f"Detailed results saved to: {OUTPUT_FILE}")
print("==============================================")
````
